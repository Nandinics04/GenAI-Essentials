### AgentRAG with LangChain

```python
import os
from typing import TypedDict, List
from langgraph.graph import StateGraph, END
from langchain_openai import ChatOpenAI
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
#set te OPENAI_API_KEY
# os.environ['OPENROUTER_API_KEY']=os.getenv('OPENROUTER_API_KEY')

#Initialise models
llm=ChatOpenAI(
    model="openrouter/free",
    api_key=os.getenv('OPENROUTER_API_KEY'),
    base_url="https://openrouter.ai/api/v1",
)
embeddings= HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

### State Defination

```python
class AgentState(TypedDict):
    question: str
    documents:List[Document]
    answer:str
    need_retrieval:bool
```

```python
## Sample Document and VectorStore

#sample document for demonstration

sample_texts=[
    "Langgraph is a framework for building multi-step workflows with LLMs and other components.",
    "RAG(Retrieval-Augmented Generation) is a technique that uses a large language model to generate text based on a combination of a large body of text and a smaller set of retrieved documents.",
    "vector databases are a type of database that store vectors, which are a type of data structure that represent a point in a high-dimensional space.",
    "Agentic systems are a type of system that use a large language model to make decisions based on a combination of a large body of text and a smaller set of retrieved documents.",
]
documents=[Document(page_content=text) for text in sample_texts]

# create a vector store
vectorstore=FAISS.from_documents(documents, embeddings)
retriever=vectorstore.as_retriever(k=3)
```

### Agents Functions

```python
def decide_retrieval(state:AgentState)->AgentState:
    """ Decide whether to retrieve documents based on the question"""
    question=state['question']

    #Simple Heuristic: if question contains certain keywords, retrieve
    retrieval_keywords=["what", "how", "explain", "describe", "tell me"]
    need_retrieval=any(keyword in question.lower() for keyword in retrieval_keywords)

    return {**state, "need_retrieval":need_retrieval}
```

```python
def retrieve_documents(state: AgentState)->AgentState:
    """Retrieve documents based on question"""
    question=state['question']
    documents=retriever.invoke(question)

    return {**state, "documents":documents}
```

```python
def generate_answer(state:AgentState)->AgentState:
    """Generate an answer using the retrived documents or direct response"""
    question=state['question']
    documents=state.get('documents',[])

    if documents:
        #RAG approach: use document as context
        context= "\n\n".join([doc.page_content for doc in documents])
        prompt=f"""Based on the following context, answer the question
        
        Context:
        {context}

        Question: {question}
        Answer:"""
    else:
        #Direct response if no retrieval
        prompt=f"""Answer the following question: {question}"""
    
    response=llm.invoke(prompt)

    return {**state, "answer":response.content}
```

### Conditional Logic


```python
def should_retrieve(state: AgentState)->str:
    """Determine the next step based on need_retrieval flag"""
    if state['need_retrieval']:
        return "retrieve"
    else:
        return "generate"
```

### Build the Graph

```python
# Create a state graph
workflow=StateGraph(AgentState)

workflow.add_node("decide", decide_retrieval)
workflow.add_node("retrieve", retrieve_documents)
workflow.add_node("generate", generate_answer)


#Set entry point
workflow.set_entry_point("decide")

#Add conditional edges
workflow.add_conditional_edges(
    "decide",
    should_retrieve,
    {
        "retrieve": "retrieve",
        "generate": "generate",
    }
)

#Add edges for retrieval and generation
workflow.add_edge("retrieve", "generate")
workflow.add_edge("generate", END)

app=workflow.compile()
app
```

### Test Agentic System

```python
def ask_question(question:str):
    """Helper function to ask a questio and get an answer"""
    initial_state={
        "question":question,
        "documents":[],
        "answer":"",
        "need_retrieval":False,
    }
    result=app.invoke(initial_state)
    return result
```

```python
question2="How does RAG work?"
result2=ask_question(question2)

print(f"Question: {question2}")
print(f"Retrieved documents: {len(result2['documents'])}")
print(f"Answer: {result2['answer']}")
print("\n"+ "="*50 + "\n")
```
