### Pydantic Basics: Creating and Using the Models

Pydantic models are the foundation of data validation in python. They use python type annotations to define the structure and validate at runtime. Here is detailed explaination of basic model creation with several examples.

```python
from pydantic import BaseModel
```

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    city:str

person=Person(name="Nandini", age=20, city="Nalgonda")
print(person)
```

```python
class Person(BaseModel):
    name: str
    age: int
    city:str

person=Person(name="John", age=30, city="New York")
print(person)
```

### Model with Optional Field 

Add Optional Fields using Python Optional type

```python
from typing import Optional
class Employee(BaseModel):
    id: int
    name: str
    department: str
    salary: Optional[str] = None
    is_active: Optional[bool]=True
```

```python
emp1=Employee(id=1,name="John",department="HR")
print(emp1)
```

Definition:
- Optional[type]: Indicates the field can be None

- Default value (=None or =True): Make the field Optional

- Required fields must still be provided

Pydantic validates the Optional field when values are provided

```python
class ClassRoom(BaseModel):
    room_number: int
    students: list[str]
    capacity: int
```

### Pydantic with Nest Models

creting complex structures with nested models

```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    zip_code: str

class Customer(BaseModel):
    name: str
    age: int
    address: Address

customer=Customer(
    name="John",
    age=30,
    address={
        "street":"123 Main St",
        "city":"Anytown",
        "zip_code":"12345"
    }
    
)
print(customer)
```

### Pydantic Fields:  Customisation and Constriants

The Field fucntion in Pydantic enhances model fields beyond the basic typ hints by allowing you to specify the validation rules, default values, aliases and more, 

```python
from pydantic import BaseModel, Field

class Item(BaseModel):
    name: str= Field(min_length=3, max_length=15)
    price: float=Field(gt=0, le=1000)
    quantity:int=Field(ge=0)


item=Item(name="Laptop", price=1500, quantity=10)
```

```python
class User(BaseModel):
    username: str=Field(..., description="Unique username for the user")
    age:int=Field(default=20, description="Age must be between 18 and 100")
    email:str=Field(...,default_factory=lambda:"user@gmail.com", alias="email_address", description="Email address")


user1=User(username="John")
print(user1)

user2=User(username="John", age=25, email="john@example.com")
print(user2)
```
